### ¿Qué es un Hamiltoniano?

En mecánica cuántica, el Hamiltoniano es el operador que representa la energía de un sistema físico.

Se suele denotar como:

$$
H
$$

Si un sistema cuántico está en el estado $|\psi\rangle$, entonces la energía esperada de ese estado es:

$$
E(\psi) = \langle \psi | H | \psi \rangle
$$

El Hamiltoniano contiene la información del sistema que queremos estudiar. Por ejemplo, puede representar:

- La energía de una molécula.
- La interacción entre espines.
- Un sistema de partículas.
- Un problema de optimización escrito en lenguaje cuántico.

En computación cuántica, normalmente escribimos el Hamiltoniano como una combinación de operadores de Pauli:

$$
H = c_0 I + c_1 X + c_2 Y + c_3 Z + \cdots
$$

donde $X$, $Y$ y $Z$ son las matrices de Pauli.

### ¿Qué es el estado base?

Los estados propios de un Hamiltoniano satisfacen:

$$
H |E_i\rangle = E_i |E_i\rangle
$$

donde:

- $|E_i\rangle$ es un estado propio del Hamiltoniano.
- $E_i$ es la energía asociada a ese estado.

El **estado base** es el estado de menor energía posible del sistema. Es decir, si las energías del Hamiltoniano son:

$$
E_0 \leq E_1 \leq E_2 \leq \cdots
$$

entonces $E_0$ es la **energía del estado base**, y $|E_0\rangle$ es el **estado base**.

Encontrar el estado base es un problema muy importante en física, química cuántica y optimización.

Por ejemplo, en química cuántica, conocer la energía del estado base de una molécula ayuda a entender su estabilidad y sus propiedades físicas.

Existe un principio muy importante, llamado el **Principio Variacional** que dice que la energía de cualquier estado siempre es mayor o igual que la energía del estado base. Pero por el momento, eso es harina de otro costal.

## Hamiltoniano de Ising

Un problema tipo Ising es un problema que se visualiza como una función de costo cuadrática con variables binarias +1 y -1. La función de costo es la siguiente:

$$f(z)=\sum _i h_is_i+ \sum _{i<j} J_{ij} s_i s_j$$

en dondi $h_i$ y $J_{i,j}$ son interacciones con campos y entre particulas que permiten que una particula tome cierto valor u otro. Un problema Ising representar encontrar el minimo **s** donde **s** es un vector binario cuyos valores son $+1$ y $-1$

Si tenemos un problema de optimización, buscaremos $\textbf{s}_{min}=s_1s_2...s_n$ tal que $f(\textbf{s}_{min})$ sea el valor mínimo. Puede ser que hayan varios pero podemos asegurar la existencia de al menos uno. Aquí, empezaremos a usar lo que sabemos de estados cuánticos para conectar estos 2 conceptos. 

Nosotros podemos construir un Hamiltoniano llamado el **Hamiltoniano de Ising** apoyandonos en las propiedades de la función que dimos anteriormente. Para ello, lo que vamos a hacer es que, vamos a cambiar en la función de Ising los $s_i$ por $\hat{Z}_i$ de tal manera que se aplique justamente el operador $\hat{Z}$ en el i-esimo qubit del sistema. 


$$H_{ising}=\sum _i h_i\hat{Z}_i+ \sum _{i<j} J_{ij} \hat{Z}_i \hat{Z}_j$$

Ahora, hay una idea clave que voy a anticiparme y que va a ser el eje, y es que *existe una forma de conectar estados base con problemas de optimización*. Vamos a ello, para entender como conectar esta idea relacionando estados cuánticos con strings


## Relación entre estados cuánticos y spines de Ising

Imagina que tenemos n qubits a nuestra disposición y que los visualizamos en la base computacional. Un estado cuantico de qubits de la base computacional podría ser:

$$|x\rangle = |x_1x_2...x_n\rangle$$

donde $x_i \in \{0,1\}$

Si queremos conectar strings con spines, podemos utilizar la siguiente conversión: $s_i=(-1)^{x_i}$. Esto lo que hace es que podemos transformar los valores que se recuperen de un estado cuántico de qubits a spines que se puedan sustituir en el interior de una función de energía de Ising

Ahora, una pregunta interesante nace. Si podemos utilizar los estados cuánticos para representar spines, ¿Podriamos utilizar las herramientas de la computación cuántica para poder encontrar el estado óptimo de minimización?

Aquí es en donde existe un teorema que nos va a permitir conectar esto:

> Teorema: Utilizando la conversión de estados cuanticos de qubits a una base computacional, el **Estado base** del Hamiltoniano de Ising asociado a una función de Ising nos devuelve la distribución de spines óptima para dicha función

### Demostración intuitiva del teorema del Hamiltoniano de Ising

Ahora, recuerda que el operador 
$\hat{Z}=\begin{pmatrix}
1 & 0 \\
0 & -1
\end{pmatrix}
$

Cumple que $\hat{Z}|0\rangle= +|0\rangle$ y que $\hat{Z}|1\rangle=-|1\rangle$. Es decir, los eigenvalores de $\hat{Z}$ son $+1$ y $-1$, tal y como es el caso de los valores de Spin del sistema. Mas aún, podemos notar incluso que $\hat{Z}|k\rangle=(-1)^k|k\rangle$ si $k$ es un número entre 0 y 1

Ahora, podemos notar que, si $|x\rangle=|x_1x_2...x_n\rangle$, entonces es facil notar que $\hat{Z}_i|x\rangle=(-1)^{x_i}|x\rangle$ de donde ademas, usando la transformación dada anteriormente, lo podemos visualizar como $\hat{Z}_i|x\rangle=s_i|x\rangle$
